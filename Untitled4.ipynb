```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import folium

# 1. تحميل البيانات (هذا هو اسم الملف الصحيح)
df = pd.read_csv('historical_automobile_sales 2.csv')

# 1.1 رسم بياني خطي: مبيعات السيارات من سنة إلى أخرى
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x='Year', y='Automobile_Sales')
plt.title('Yearly Automobile Sales')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.show()

# 1.2 رسم بياني خطي: اتجاهات الإنفاق على الإعلانات خلال فترة الركود
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x='Year', y='Advertising_Expenditure', hue='Recession')
plt.title('Advertising Expenditure Trends during Recession')
plt.show()

# 1.3 رسم بياني عمودي: متوسط المبيعات حسب نوع السيارة (فترة الركود مقابل عدم الركود)
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='Vehicle_Type', y='Automobile_Sales', hue='Recession')
plt.title('Average Sales by Vehicle Type: Recession vs Non-Recession')
plt.xticks(rotation=45)
plt.show()

# 1.4 رسومات فرعية: مقارنة التغيرات في الناتج المحلي الإجمالي (GDP) مع المبيعات
fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sns.lineplot(data=df, x='Year', y='GDP', ax=axes[0])
axes[0].set_title('GDP Trends')
sns.lineplot(data=df, x='Year', y='Automobile_Sales', ax=axes[1])
axes[1].set_title('Automobile Sales Trends')
plt.show()

# 1.5 رسم بياني قاعي: تأثير الموسم على مبيعات السيارات
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Month', y='Automobile_Sales')
plt.title('Seasonal Impact on Automobile Sales')
plt.show()

# 1.6 رسم بياني مبعثر: العلاقة بين متوسط سعر السيارة وحجم المبيعات خلال الركود
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df[df['Recession'] == 1], x='Price', y='Automobile_Sales')
plt.title('Price vs Sales Volume during Recession')
plt.show()

# 1.7 رسم بياني دائري: نسبة الإنفاق على الإعلانات (فترة الركود مقابل عدم الركود)
advertising_data = df.groupby('Recession')['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 8))
plt.pie(advertising_data, labels=['No Recession', 'Recession'], autopct='%1.1f%%')
plt.title('Advertising Expenditure Share: Recession vs No Recession')
plt.show()

# 1.8 رسم بياني دائري: إجمالي الإنفاق على الإعلانات لكل نوع سيارة
ad_exp_type = df.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()
plt.figure(figsize=(8, 8))
plt.pie(ad_exp_type, labels=ad_exp_type.index, autopct='%1.1f%%')
plt.title('Advertising Expenditure by Vehicle Type')
plt.show()

# 1.9 رسم بياني خطي: تأثير معدل البطالة على المبيعات خلال الركود
plt.figure(figsize=(10, 6))
sns.lineplot(data=df[df['Recession'] == 1], x='unemployment_rate', y='Automobile_Sales')
plt.title('Impact of Unemployment Rate on Sales during Recession')
plt.show()
```
